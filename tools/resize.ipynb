```python
import os
from pathlib import Path
from PIL import Image
```

```python
def load_image(img_path: str) -> Image:
    try:
        with Image.open(img_path) as img:
            # Force load image data into memory before the file closes
            img.load()
            print(f"Loaded: {img_path.name} | Dimensions: {img.size} | Mode: {img.mode}")
            
            # Example processing:
            # resized = img.resize((300, 300))
    except Exception as e:
        print(f"Failed to read {img_path.name}: {e}")
    return img

def resize_and_save(img: Image, output_path: str, size: tuple = (300, 300)):
    # Convert to RGB if image has an alpha channel (PNG/RGBA) to avoid JPEG save errors
    if img.mode in ("RGBA", "P"):
        img = img.convert("RGB")
    
    # Resize using high-quality downsampling (LANCZOS)
    resized_img = img.resize(size, Image.Resampling.LANCZOS)
    if len(output_path) > 0:
        # Save as JPEG
        resized_img.save(output_path, format="JPEG", quality=90)
    
    return resized_img
```

```python
# Resize for THINGS DATASET
output_dir = Path("../images/resized/")
categories = [
    "clothing",
    "food",
    "fruit",
    "music",
    "appliance",
    "furniture",
    "vehicle",
    "land_animal",
    "office_supply",
]
for fi, c in enumerate(categories):
    image_dir = Path(f"../images/{c}/")
    # Use *.png (case-sensitive on Linux) or rglob for recursive search
    for i, img_path in enumerate(image_dir.glob(f"*.jpg")):
        img = load_image(img_path)
        # Create output directory if it does not exist
        out_folder = output_dir / f"img{fi+2}"
        out_folder.mkdir(parents=True, exist_ok=True)

        resized_img = resize_and_save(img, f"{out_folder}/image{i+1}.jpg")
```
